## `while` loops
```python
while <condition>:
    statement
    statement
    ...
    statement
```

<br>

- `<condition>` evaluates to a Boolean (`True` or `False`)

    1. if `<condition>` is `True`, then execute all statements inside the `while` block
    2. evaluate `<condition>` again
    3. repeat until `<condition>` is `False`

### Example

````python
%%script echo skipping

n = input("You are in the Lost Forest. Go Left or Right?: ")
while n == "Right":
    n = input("You are in the Lost Forest. Go Left or Right?: ")
print("You are out of the Lost Forest.")
````

## `for` loops
```python
for <variable> in range(<number>):
    statement
    statement
    ...
    statement
```

<br>

- each time through the loop, `<variable>` takes some value
- `<variable>` starts from the smallest value (usually, `1`)
- next time, `<variable> = <variable> + 1`

### The `range()` function
The `range(start, stop, step)` function returns a sequence of numbers, starting from `start`, and increments by `step`, and stops at `stop-1`.\
Keep in mind the following caveats:
- `stop` is mandatory
- `step` is optional: if not provided, it defaults to `1`
- `start` is optional: if not provided, it defaults to `0`

Examples:

````python
mysum = 0
for i in range(7, 10):
    mysum = mysum + i
print(mysum)
````

````python
mysum = 0
for i in range(5, 11, 2):
    mysum = mysum + i
print(mysum)
````

## `for` vs. `while` loops
Iterate through sequence of numbers in `[0, 5)`:

- With a `while` loop:

````python
n = 0
while n < 4:
    print(n)
    n = n + 1
````

- With a `for` loop:

````python
for n in range(4):
    print(n)
````

### Main difference

<br>

<div style="display: flex; justify-content: space-around;">
  <div>
    <b>for loops</b>
    <ul>
      <li><b style="color:red">known</b> number of iterations</li>
      <li>can <b style="color:red">end early</b> via <code>break</code></li>
      <li>(natively) uses a <b style="color:red">counter</b></li>
      <li>you <b style="color:red">can rewrite</b> a <code>for</code> loop using a <code>while</code> loop</li>
    </ul>
  </div>
  <div>
    <b>while loops</b>
    <ul>
      <li><b style="color:red">unbounded</b> number of iterations</li>
      <li>can <b style="color:red">end early</b> via <code>break</code></li>
      <li>you can use a <b style="color:red">counter</b>, but you must initialise it before the loop and increment it inside the loop</li>
      <li>you <b style="color:red">may not be able to rewrite</b> a <code>while</code> loop using a <code>for</code> loop</li>
    </ul>
  </div>

## More on `for` loops
`for` loops can be used to iterate over other data structures (e.g., lists or dictionaries*).

Example: print all functions in the `math` module that contain the word '`log`';

## Nested Loops
We know that branching (i.e., `if`) statements can be arbitrarily nested. The same holds for loops.

Example:

````python
mysum = 0

for i in range(3):
    for x in range(2):
        mysum += i

print(mysum)
````

<div class="alert alert-block alert-info"><b>RECALL</b>: <code>mysum += i</code> is the same as <code>mysum = mysum + i</code></div>

## `break` statement

The `break` statement:
- immediately exits whatever loop is it in
- all further statements in the block are skipped
- exits only the innermost loop
- usually triggered at given condition

<br>

<br>

```python
while <condition_1>:
    while <condition_2>:
        <statement_a>
        break           # exits inner loop
        <statement_b>   # never executed
    <statement_c>
```

### Example

````python
mysum = 0

for i in range(5, 11, 2):
    mysum += i
    if mysum == 5:
        break
    mysum += i

print(mysum)
````


What's going on here?

````python
mysum = 0

for i in range(30):
    for n in range(20):
        if n > 2:
            break
        mysum += i
    if i>1:
        break

print(mysum)
````

### Exercise 1: Coin flipping

- Ask the user how many coin tosses have to be performed
- Keep track of the number of times we get heads and tails
- We assume a *fair coin* (i.e., unbiased 50% change of getting head and tail at each toss)

<br>

<br>
<div class="alert alert-block alert-info"><b>HINT</b>: You may use the <code>random.randint()</code> function to simulate a single coin toss.</div>

<br>

````python
import random

tail = 0
head = 0

n = 3 # number of coin tosses
# n = int(input("How many coin tosses? "))
for i in range(n):
    x = random.randint(1, 100)
    if x > 50:
        head = head + 1
    else:
        tail = tail + 1

print("Number of heads: ", head)
print("Number of tails: ", tail)
````

### Exercise 2: Sum of sequence

Write a function named `rolling_sum()` that:
- repeatedly asks the user to insert a number
- we assume integer numbers
- keeps track of the sum of the numbers inserted so far
- if the user inserts 0, then exit and return the sum

<br>

````python
%%script echo skipping

def rolling_sum():
    this_number = int(input("Insert a number: "))
    my_sum = this_number

    while this_number != 0:
        this_number = int(input("Insert a number: "))
        my_sum = my_sum + this_number

    return my_sum

print("Total sum is: ", rolling_sum())
````

### Exercise 3: Odd or even?

Write a function named `odd_or_even()` that:
- accepts an integer number, say n
- prints the number of odd numbers in range [1,n] (i.e., both included)
- prints the number of even numbers in range [1,n] (i.e., both included)

<br>

````python
def odd_or_even(n):
    odd = 0
    even = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            even += 1
        else:
            odd += 1

    print("Number of odd elements: ", odd)
    print("Number of even elements: ", even)

odd_or_even(10)
````

### Exercise 4: The Fibonacci Sequence

Write a function named `fibonacci1()` that:
- accepts an integer number, say n
- prints the first n elements of the Fibonacci sequence.

<br>

<br>
<div class="alert alert-block alert-info"><b>Definition</b>: Each number in the Fibonacci sequence is given by the sum of the two preceding ones.\The Fibonacci sequence starts with two 'seed' numbers <code>0</code> and <code>1</code>) and then the proper sequence starts: <code>0, 1, <b style="color:blue">1, 2, 3, 5, 8, 13, 21, ...</b></code></div>

<br>

````python
def fibonacci1(n):
    second_to_last = 0
    last = 1

    print(second_to_last)
    print(last)

    for i in range(n - 2):
        new = last + second_to_last
        print(new)
        second_to_last = last
        last = new

fibonacci1(10)
````

### Exercise 5: Multiples

Write a function named `multiples()` that:
- accepts an integer number, say n
- iterates the integers in [1, n]
- for multiples of 3, print `'Mult3'` instead of the number
- for multiples of 5, print `'Mult5'` instead of the number
- for multiples of 5 and 3, print `'Mult35'` instead of the number

<br>

````python
def multiples(n):
    for num in range(1, n + 1):
        if num % 3 == 0 and num % 5 == 0:
            print("Mult35")
        elif num % 3 == 0:
            print("Mult3")
        elif num % 5 == 0:
            print("Mult5")
        else:
            print(num)

multiples(15)
````

### Exercise 6: Print a subset

Write a function named `print_subset()` that:
- accepts an integer number, say n
- accepts an integer number, say m <= n
- scans all integer numbers in range [1, n], but prints all numbers in range [1, m]

#### Solution 1

````python
def print_subset(n, m):
    for num in range(1, n + 1):
        if num <= m:
            print(num)

print_subset(5, 3)
````

#### Solution 2

````python
def print_subset(n, m):
    for num in range(1, n + 1):
        print(num)
        if num == m:
            break

print_subset(5, 3)
````

### Exercise 7: Primality test

Write a function named `smallerPrimes()` that:
- accepts a positive integer number, say n
- prints all positive integer prime numbers smaller than n

<br>

<br>
<div class="alert alert-block alert-info"><b>Recall</b>: A *prime* number is a natural number greater than 1 that has no positive divisors other than 1 and itself.</div>

<br>
<div class="alert alert-block alert-info"><b>Hint</b>: The simplest primality test aims at checking whether n is evenly divisible (i.e., division with no remainder) by any number in range [2, n/2]. If so, the number is not prime.</div>

<br>

````python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num/2) + 1):
        if num % i == 0:
            return False
    return True

def smallerPrimes(n):
    for i in range(2, n):
        if is_prime(i):
            print(i)

smallerPrimes(20)
````

### Exercise 8: Pythagorean triples

Write a function named `pythagoreanTriples()` that:
- accepts an integer positive number, say n
- prints all x and y pairs such that x<sup>2</sup> + y<sup>2</sup> = n<sup>2</sup>

<br>

<br>
<div class="alert alert-block alert-info"><b>Recall</b>: The Pythagorean triple consists of three positive integers x, y and n, with x,y &le; n, such that: x<sup>2</sup> + y<sup>2</sup> = n<sup>2</sup></div>

<br>

````python
def pythagoreanTriples(n):
    for x in range(1, n):
        for y in range(1, n):
            if x**2 + y**2 == n**2:
                print(x, y)

pythagoreanTriples(5)
````
