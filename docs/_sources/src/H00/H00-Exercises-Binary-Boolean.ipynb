## Binary Number Exercises

### Decimal 387 to Binary

Convert the decimal number **387** to binary:

$387 \div 2 = 193$ (rem 1)

$193 \div 2 = 96$ (rem 1)

$96 \div 2 = 48$ (rem 0)

$48 \div 2 = 24$ (rem 0)

$24 \div 2 = 12$ (rem 0)

$12 \div 2 = 6$ (rem 0)

$6 \div 2 = 3$ (rem 0)

$3 \div 2 = 1$ (rem 1)

$1 \div 2 = 0$ (rem 1)

Result: $(110000011)_2$

### Binary 110110101011 to Decimal

Convert the binary number **110110101011** to decimal:

$$ (110110101011)_2 = 1 \cdot 2^{11} + 1 \cdot 2^{10} + 0 \cdot 2^9 + 1 \cdot 2^8 + 1 \cdot 2^7 + 0 \cdot 2^6 + 1 \cdot 2^5 + 0 \cdot 2^4 + 1 \cdot 2^3 + 0 \cdot 2^2 + 1 \cdot 2^1 + 1 \cdot 2^0 $$
$$ = 2048 + 1024 + 0 + 256 + 128 + 0 + 32 + 0 + 8 + 0 + 2 + 1 = (3499)_{10} $$

### Decimal 2048 to Hexadecimal

Convert the decimal number **2048** to hexadecimal:

$2048 \div 16 = 128$ remainder $0$

$128 \div 16 = 8$ remainder $0$

$8 \div 16 = 0$ remainder $8$

Reading the remainders from bottom to top: $800$

Therefore, $2048_{10} = 800_{16}$

### Hex BADC0DE to Binary

Convert the hexadecimal number **BADC0DE** to binary:

Convert each hexadecimal digit to its 4-bit binary equivalent:

- $B_{16} = {1011}_2$
- $A_{16} = {1010}_2$
- $D_{16} = {1101}_2$
- $C_{16} = {1100}_2$
- $0_{16} = {0000}_2$
- $D_{16} = {1101}_2$
- $E_{16} = {1110}_2$

Therefore: $BADC0DE_{16} = 1011101011011100000011011110_2$
