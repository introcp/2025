## Sockets: The Foundation

A **socket** is an endpoint for sending or receiving data across a network.

Think of it as a "phone connection" between two computers:
- One computer (server) listens for incoming connections
- Another computer (client) initiates the connection
- Once connected, they can exchange data

Python provides the `socket` module for low-level network programming.

<div class="alert alert-block alert-info">
<b>NOTE</b>: Sockets are the foundation of all network communication. HTTP, email, file transfer - they all use sockets under the hood!
</div>

### Simple Socket Example

````python
import socket

# Create a socket object
mysock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Connect to a web server on port 80 (HTTP)
mysock.connect(('data.pr4e.org', 80))

# Send an HTTP request
cmd = 'GET http://data.pr4e.org/romeo.txt HTTP/1.0\r\n\r\n'.encode()
mysock.send(cmd)
````

````python
while True: 
    data = mysock.recv(512) # Receive data
    if len(data) < 1:
        break
    print(data.decode(), end='')

mysock.close()
````

<div class="alert alert-block alert-info">
<b>NOTE</b>: In the code above:
<ul>
    <li><code>AF_INET</code> means we're using IPv4 addressing</li>
    <li><code>SOCK_STREAM</code> means we're using TCP (reliable, ordered communication)</li>
    <li>Port 80 is the default port for HTTP</li>
    <li>The <code>.encode()</code> method converts strings to bytes (sockets work with bytes)</li>
    <li>The <code>.decode()</code> method converts bytes to strings</li>
</ul>
</div>

### Making HTTP GET Request

The most common HTTP operation is GET - fetching data from a server. The same functionality as our socket example, but much simpler:

````python

import requests

# Make a GET request
response = requests.get('http://data.pr4e.org/romeo.txt')

# The response object contains:
print(response.status_code)  # HTTP status code (200 = success)
print(response.text[:50])    # Response body as text
print(response.headers)      # Response headers
````

## JSON: The Universal Data Format

JSON (JavaScript Object Notation) is a lightweight data-interchange format that has become the standard for exchanging data on the web.

Why is JSON so popular?
- It's human-readable
- It's language-independent
- It can represent complex data structures
- It's easy for machines to parse and generate

### JSON Syntax

JSON data types map to Python data types:

| JSON | Python |
|------|--------|
| `string` | `str` |
| `number` | `int` or `float` |
| `boolean` | `bool` |
| `null` | `None` |
| `array` | `list` |
| `object` | `dict` |

Example of JSON data:

````python
{
    "name": "Alice",
    "age": 30,
    "is_student": false,
    "courses": ["Math", "Physics", "CS"],
    "address": {
        "street": "123 Main St",
        "city": "Rome"
    }
}
````

### Working with JSON in Python

Python's `json` module (built-in!) provides functions to convert between JSON strings and Python objects:

````python
import json

# Python dictionary
data = {
    "name": "Alice",
    "age": 30,
    "courses": ["Math", "Physics"]
}

# Convert Python dict to JSON string
json_string = json.dumps(data)
print(json_string)

# Convert JSON string back to Python dict
data_again = json.loads(json_string)
print(data_again)
````

### Reading and Writing JSON Files

````python
import json

# Write JSON to file
data = {"name": "Bob", "age": 25}
with open("person.json", "w") as f:
    json.dump(data, f)

# Read JSON from file
with open("person.json", "r") as f:
    loaded_data = json.load(f)
    print(loaded_data)  # {'name': 'Bob', 'age': 25}
````

<div class="alert alert-block alert-info">
<b>NOTE</b>: Remember the key functions:<br>
<ul>
    <li><code>json.dumps()</code>: Python object → JSON string (dumps = "dump string")</li>
    <li><code>json.loads()</code>: JSON string → Python object (loads = "load string")</li>
    <li><code>json.dump()</code>: Python object → JSON file</li>
    <li><code>json.load()</code>: JSON file → Python object</li>
</ul>
</div>

## REST APIs: Interacting with Web Services

REST (Representational State Transfer) is an architectural style for designing web services.

Most modern web APIs follow REST principles:
- Resources are identified by URLs
- HTTP methods (GET, POST, PUT, DELETE) specify the operation
- Data is exchanged in JSON format
- Each request is independent (stateless)


### Fetching Data from a Public API

Let's start with a simple public API that provides interesting data:

````python
import requests

# Fetch a random joke from a public API
response = requests.get('https://official-joke-api.appspot.com/random_joke')

# Check if successful
if response.status_code == 200:
    joke = response.json()  # Automatically parses JSON!
    print(f"Setup: {joke['setup']}")
    print(f"Punchline: {joke['punchline']}")
else:
    print(f"Error: {response.status_code}")
````

<div class="alert alert-block alert-success">
<b>HINT</b>: The <code>requests</code> library makes it incredibly easy to work with JSON APIs! The <code>.json()</code> method automatically converts the JSON response into a Python dictionary.
</div>

### GET Requests with Parameters

Parameters (also called query parameters or query strings) are used when you need to customize or filter the data you're requesting from an API. Think of them as "options" or "settings" for your request.

````python
import requests

def get_weather(city, units='metric'):
    """Get weather data for a city using Open-Meteo API"""
    
    # Step 1: Get coordinates for the city
    geocoding_url = 'https://geocoding-api.open-meteo.com/v1/search'
    geocoding_params = {
        'name': city,
        'count': 1
    }
    
    geo_response = requests.get(geocoding_url, params=geocoding_params)
    geo_data = geo_response.json()
    
    # Check if city was found
    if 'results' not in geo_data or len(geo_data['results']) == 0:
        return None
    
    # Extract coordinates
    latitude = geo_data['results'][0]['latitude']
    longitude = geo_data['results'][0]['longitude']
    city_name = geo_data['results'][0]['name']
    
    # Step 2: Get weather data
    weather_url = 'https://api.open-meteo.com/v1/forecast'
    temp_unit = 'celsius' if units == 'metric' else 'fahrenheit'
    
    weather_params = {
        'latitude': latitude,
        'longitude': longitude,
        'current_weather': 'true',
        'temperature_unit': temp_unit
    }
    
    weather_response = requests.get(weather_url, params=weather_params)
    weather_data = weather_response.json()
    
    # Extract temperature
    temperature = weather_data['current_weather']['temperature']
    
    return {
        'city': city_name,
        'temperature': temperature,
        'units': units
    }

# Test the function
weather = get_weather('Rome')
if weather:
    print(f"City: {weather['city']}")
    print(f"Temperature: {weather['temperature']}°C")
    print(f"Units: {weather['units']}")
````

**When are parameters used in GET requests?**

Common use cases for parameters:

1. **Filtering data** - Narrow down results (e.g., searching for a specific city: `name=Rome`)
2. **Customizing the response** - Control how data is returned (e.g., choosing temperature units: `temperature_unit=celsius`)
3. **Specifying what to retrieve** - Tell the API exactly what data you want (e.g., `current_weather=true`)
4. **Providing identifiers** - Pass coordinates, IDs, or other identifiers (e.g., `latitude=41.9`, `longitude=12.5`)
5. **Pagination** - Request specific chunks of data (e.g., `page=2`, `limit=10`)
6. **Authentication** - Include API keys or tokens (e.g., `api_key=YOUR_KEY`)

Without parameters, you'd get a generic or default response. With parameters, you can ask for exactly what you need!

## POST Requests: Sending Data

While GET requests fetch data, POST requests send data to servers.

### Creating Data with POST

````python
import requests

def create_bin(data):
    """Create a JSON storage bin using JSONBin.io API"""
    url = 'https://api.jsonbin.io/v3/b'
    
    # Set headers
    headers = {
        'Content-Type': 'application/json'
    }
    
    try:
        # Make POST request with JSON data
        response = requests.post(url, json=data, headers=headers)
        
        # Check if successful
        success = response.status_code in [200, 201]
        
        if success:
            # Parse response
            response_data = response.json()
            
            # Extract bin ID
            bin_id = response_data['metadata']['id']
            
            # Build URL
            bin_url = f'https://api.jsonbin.io/v3/b/{bin_id}'
            
            return {
                'success': True,
                'bin_id': bin_id,
                'url': bin_url
            }
        else:
            return {
                'success': False,
                'bin_id': None,
                'url': None
            }
    
    except:
        return {
            'success': False,
            'bin_id': None,
            'url': None
        }

# Test the function
test_data = {
    'name': 'John Doe',
    'age': 30,
    'city': 'Rome'
}

result = create_bin(test_data)
print(f"Success: {result['success']}")
if result['success']:
    print(f"Bin ID: {result['bin_id']}")
    print(f"URL: {result['url']}")
````

<div class="alert alert-block alert-info">
<b>NOTE</b>: When using <code>requests.post()</code> with the <code>json</code> parameter:
<ul>
    <li>The data is automatically serialized to JSON</li>
    <li>The Content-Type header is automatically set to application/json</li>
    <li>You don't need to call <code>json.dumps()</code> manually</li>
</ul>

## Error Handling

Robust applications must handle various types of errors when working with APIs.

### Example 8: Comprehensive Error Handling

````python
import requests

def get_weather_safe(city, api_key):
    """Fetch weather data with comprehensive error handling"""
    url = 'https://api.openweathermap.org/data/2.5/weather'
    
    params = {
        'q': city,
        'appid': api_key,
        'units': 'metric'
    }
    
    try:
        # Make GET request with timeout
        response = requests.get(url, params=params, timeout=10)
        
        # Get status code
        status_code = response.status_code
        
        # Handle different status codes
        if status_code == 200:
            # Success - parse data
            try:
                data = response.json()
                return {
                    'success': True,
                    'error': None,
                    'status_code': status_code,
                    'data': {
                        'city': data['name'],
                        'temperature': data['main']['temp'],
                        'description': data['weather'][0]['description']
                    }
                }
            except (KeyError, ValueError):
                return {
                    'success': False,
                    'error': 'Failed to parse API response',
                    'status_code': status_code,
                    'data': None
                }
        
        elif status_code == 401:
            return {
                'success': False,
                'error': 'Invalid API key. Please check your credentials.',
                'status_code': status_code,
                'data': None
            }
        
        elif status_code == 404:
            return {
                'success': False,
                'error': f'City "{city}" not found. Please check the spelling.',
                'status_code': status_code,
                'data': None
            }
        
        elif status_code == 429:
            return {
                'success': False,
                'error': 'Too many requests. Please wait and try again later.',
                'status_code': status_code,
                'data': None
            }
        
        elif status_code >= 500:
            return {
                'success': False,
                'error': 'Server error. Please try again later.',
                'status_code': status_code,
                'data': None
            }
        
        else:
            return {
                'success': False,
                'error': f'Unexpected error (status code: {status_code})',
                'status_code': status_code,
                'data': None
            }
    
    except requests.exceptions.Timeout:
        return {
            'success': False,
            'error': 'Request timeout. Please check your internet connection.',
            'status_code': None,
            'data': None
        }
    
    except requests.exceptions.ConnectionError:
        return {
            'success': False,
            'error': 'Connection error. Please check your internet connection.',
            'status_code': None,
            'data': None
        }
    
    except Exception as e:
        return {
            'success': False,
            'error': f'Unexpected error: {str(e)}',
            'status_code': None,
            'data': None
        }

# Test with invalid API key (401 error)
result = get_weather_safe('London', 'invalid_api_key_12345')
print(f"Success: {result['success']}")
print(f"Status Code: {result['status_code']}")
print(f"Error: {result['error']}")
````

<div class="alert alert-block alert-success">
<b>BEST PRACTICES</b> for error handling:<br>
<ul>
    <li>Always check status codes before processing data</li>
    <li>Provide user-friendly error messages</li>
    <li>Handle network errors (timeout, connection failures)</li>
    <li>Use try-except blocks for JSON parsing</li>
    <li>Set timeouts for requests to prevent hanging</li>
</ul>

## API Authentication

Many APIs require authentication to prevent abuse and track usage.

### API Key Authentication

````python
import requests

def get_weather_with_api_key(city, api_key):
    """Get weather data using OpenWeatherMap API with authentication"""
    url = 'https://api.openweathermap.org/data/2.5/weather'
    
    # Set up parameters including API key
    params = {
        'q': city,
        'appid': api_key,  # API key authentication
        'units': 'metric'
    }
    
    try:
        # Make GET request with API key in parameters
        response = requests.get(url, params=params)
        
        # Check if request was successful
        if response.status_code != 200:
            return None
        
        # Parse JSON response
        data = response.json()
        
        # Extract weather information
        return {
            'city': data['name'],
            'temperature': data['main']['temp'],
            'description': data['weather'][0]['description'],
            'humidity': data['main']['humidity']
        }
    
    except:
        return None

API_KEY = 'YOUR_OPENWEATHER_API_KEY_HERE'

weather = get_weather_with_api_key('Rome', API_KEY)
if weather:
    print(f"City: {weather['city']}")
    print(f"Temperature: {weather['temperature']}°C")
    print(f"Description: {weather['description']}")
    print(f"Humidity: {weather['humidity']}%")
````

<b>SECURITY NOTE</b>: Never hardcode API keys in your code!

## SDKs: Even Higher-Level Abstractions

Many popular web services provide **SDKs** (Software Development Kits) that wrap their APIs.

**Why use SDKs instead of direct HTTP requests?**
- **Simpler code**: No need to construct URLs, headers, or handle raw JSON
- **Type safety**: Many SDKs provide type hints and autocomplete
- **Better error handling**: SDKs often provide meaningful exceptions
- **Built-in features**: Automatic retries, rate limiting, pagination
- **Easier to use**: You work with objects and methods instead of HTTP verbs

Think of it as another level of abstraction:
1. **Sockets** (lowest level): You handle everything manually
2. **requests library**: Simplifies HTTP operations
3. **SDKs** (highest level): Service-specific, easy-to-use interfaces

### Comparing with Raw Requests

To appreciate why the SDK is useful, let's see how you'd do the **same thing** using the `requests` library directly:

````python
import requests
import json

# You have to manually construct the URL
url = "https://api.gpt4-all.xyz/v1/chat/completions"

# You have to manually create the headers with authentication
headers = {
    "Authorization": f"Bearer API_KEY_HERE",
    "Content-Type": "application/json"
}

# You have to manually structure the request body as a dictionary
data = {
    "model": "gpt-4o-mini",
    "messages": [{"role": "user", "content": "List the best 5 books for a python developer"}],
    "stream": False
}
````

````python
# Make the POST request
response = requests.post(url, headers=headers, json=data)

# Check if the request was successful
if response.status_code == 200:
    # You have to manually parse the JSON response
    result = response.json()
    # You have to know the exact structure of the response
    print(result["choices"][0]["message"]["content"])
else:
    print(f"Error: {response.status_code}")
    print(response.text)
````

### What the SDK Does for You

Notice the differences? The SDK handles all these details automatically:

| Task | Raw Requests | SDK |

| **Authentication** | Create headers dictionary | Pass `api_key` parameter |
| **Request Formatting** | Build JSON dictionary manually | Use method parameters |
| **Error Handling** | Check status codes manually | Raises meaningful exceptions |
| **Response Parsing** | Navigate raw JSON dictionary | Access response object properties |

## Common HTTP Status Codes

Understanding HTTP status codes is essential for debugging API issues:

### Success Codes (2xx)
- **200 OK**: Request succeeded
- **201 Created**: Resource successfully created
- **204 No Content**: Success, but no content to return



### Client Error Codes (4xx)
- **400 Bad Request**: Invalid request syntax
- **401 Unauthorized**: Authentication required or failed
- **403 Forbidden**: Server understood request but refuses
- **404 Not Found**: Resource doesn't exist
- **429 Too Many Requests**: Rate limit exceeded

### Server Error Codes (5xx)
- **500 Internal Server Error**: Generic server error
- **502 Bad Gateway**: Invalid response from upstream server
- **503 Service Unavailable**: Server temporarily unavailable
